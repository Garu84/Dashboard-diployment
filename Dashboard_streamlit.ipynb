```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('university_student_data.csv')
print(df.head())
print(df.info())
print(df.describe())
```

**Columns**

**years:** Academic year in which the data was recorded

**term:** Academic semester or term (Spring or Fall)

**aplications:** Total number of student

**applications:** received by the university

**admitted:** Number of students accepted by the university

**enrolled:** Number of students who officially enrolled after being admitted

**retentin rate:** Percentage of students who continued their studies at the university.

**student satisfaction:** Percentage representing student satisfaction levels based on surveys or evaluations

**Engineering Enrolled:** Number of students enrolled in Engineering programs

**Business Enrolled:** Number of students enrolled in Business programs

**Arts Enrolled:** Number of students enrolled in Arts programs

**Science Enrolled:** Number of students enrolled in Science programs



```python
#Retention rate over time
retention = df.groupby('Year')['Retention Rate (%)'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(data=retention, x='Year', y='Retention Rate (%)', marker='o')
plt.title('Retention Rate Over Time')
plt.show()


#Student satisfation scores by year
satisfaction = df.groupby('Year')['Student Satisfaction (%)'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(data=satisfaction, x='Year', y='Student Satisfaction (%)')
plt.title('Student Satisfaction by Year')
plt.show()

#Comparison betwen spring vs fall tems
term_compare = df.groupby('Term')['Enrolled'].sum()

plt.figure(figsize=(7,7))

plt.pie(
    term_compare,
    labels=term_compare.index,
    autopct='%1.1f%%')

plt.title('Spring vs Fall Enrollment Comparison')

plt.show()
```
